#### Checking account balance

```python
print("Negative balances:", (df["balance"] < 0).sum())
print("Zero balances:", (df["balance"] == 0).sum())
print("Positive balances:", (df["balance"] > 0).sum())
```

#### Checking campaign contacts

```python
print("Campaign below 1:", (df["campaign"] < 1).sum())
print("Campaign above 10:", (df["campaign"] > 10).sum())

print(
    df.loc[df["campaign"] > 10,
           ["age", "job", "campaign", "deposit"]]
      .sort_values("campaign", ascending=False)
      .head(10)
)
```

#### Understanding pdays

```python
print(df["pdays"].value_counts().head(10))

print("Never previously contacted:", (df["pdays"] == -1).sum())
print("Previously contacted:", (df["pdays"] != -1).sum())
```

##### In this dataset, pdays = -1 means the customer was not previously contacted in an earlier campaign. 
##### It is a special code, not an ordinary negative number.

#### Creating useful derived columns

#### Customer age group

```python

df["age_group"] = pd.cut(
    df["age"],
    bins=[17, 25, 35, 45, 55, 65, float("inf")],
    labels=[
        "18-25",
        "26-35",
        "36-45",
        "46-55",
        "56-65",
        "66+"
    ]
)

print(df[["age", "age_group"]].head(10))
```

#### Previous contact status

```python
df["previously_contacted"] = df["pdays"].apply(
    lambda x: "No" if x == -1 else "Yes"
)

print(df["previously_contacted"].value_counts())
```

#### Balance category

```python
df["balance_category"] = np.select(
    [
        df["balance"] < 0,
        df["balance"] == 0,
        df["balance"] <= 1000,
        df["balance"] <= 5000
    ],
    [
        "Negative",
        "Zero",
        "Low",
        "Medium"
    ],
    default="High"
)

print(df["balance_category"].value_counts())
```

#### Campaign contact group

```python
df["campaign_group"] = pd.cut(
    df["campaign"],
    bins=[0, 1, 3, 5, float("inf")],
    labels=[
        "1 Contact",
        "2-3 Contacts",
        "4-5 Contacts",
        "6+ Contacts"
    ]
)

print(df["campaign_group"].value_counts())
```

#### EDA

##### Customers quantity in each deposit category and percentage distribution

```python
deposit_summary = pd.DataFrame({
    "Customer_Count": df["deposit"].value_counts(),
    "Percentage": (
        df["deposit"].value_counts(normalize=True) * 100
    ).round(2)
})

deposit_summary.index.name = "Deposit_Status"

print(deposit_summary)
```

#### Creating a KPI variable

```python
total_customers = len(df)

subscribed_customers = (df["deposit"] == "yes").sum()

subscription_rate = (
    subscribed_customers / total_customers
) * 100

print("Total Customers:", total_customers)
print("Subscribed Customers:", subscribed_customers)
print("Subscription Rate:", round(subscription_rate, 2), "%")
```

##### Approximately 47 out of every 100 customers subscribed to a term deposit in this dataset. The remaining 53 did not.

#### Which customer groups have higher term-deposit subscription rates?

```python
job_analysis = df.groupby("job", observed=True).agg(
    Total_Customers=("deposit", "count"),
    Subscribed=("deposit", lambda x: (x == "yes").sum())
)

job_analysis["Subscription_Rate"] = (
    job_analysis["Subscribed"]
    / job_analysis["Total_Customers"]
    * 100
).round(2)

job_analysis = job_analysis.sort_values(
    "Subscription_Rate",
    ascending=False
)

print(job_analysis)
```

#### Analyzing subscription by marital status

```python
marital_analysis = df.groupby("marital", observed=True).agg(
    Total_Customers=("deposit", "count"),
    Subscribed=("deposit", lambda x: (x == "yes").sum())
)

marital_analysis["Subscription_Rate"] = (
    marital_analysis["Subscribed"]
    / marital_analysis["Total_Customers"]
    * 100
).round(2)

print(
    marital_analysis.sort_values(
        "Subscription_Rate",
        ascending=False
    )
)
```

#### Analyzing subscription by education

```python
education_analysis = df.groupby("education", observed=True).agg(
    Total_Customers=("deposit", "count"),
    Subscribed=("deposit", lambda x: (x == "yes").sum())
)

education_analysis["Subscription_Rate"] = (
    education_analysis["Subscribed"]
    / education_analysis["Total_Customers"]
    * 100
).round(2)

print(
    education_analysis.sort_values(
        "Subscription_Rate",
        ascending=False
    )
)
```

#### Analyzing subscription by contact type

```python
contact_analysis = df.groupby("contact", observed=True).agg(
    Total_Customers=("deposit", "count"),
    Subscribed=("deposit", lambda x: (x == "yes").sum())
)

contact_analysis["Subscription_Rate"] = (
    contact_analysis["Subscribed"]
    / contact_analysis["Total_Customers"]
    * 100
).round(2)

print(
    contact_analysis.sort_values(
        "Subscription_Rate",
        ascending=False
    )
)
```
